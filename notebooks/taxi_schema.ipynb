```python
import pyspark
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder \
    .master("local[*]") \
    .appName('taxi_preparation') \
    .getOrCreate()
```

```python
# Loop through years and taxi types
taxi_types = ['green', 'yellow']
years = [2020, 2021]

for taxi in taxi_types:
    for year in years:
        for month in range(1, 13):
            print(f'Processing data for {taxi} {year}/{month:02d}')

            input_path = f'../data/raw/{taxi}/{year}/{month:02d}/*.parquet'
            output_path = f'../data/pq/{taxi}/{year}/{month:02d}/'

            try:
                # 1. Read the raw parquet (Schema is automatically inferred!)
                df = spark.read.parquet(input_path)

                # 2. Repartition to 4 chunks and write back to disk
                df \
                    .repartition(4) \
                    .write.mode('overwrite').parquet(output_path)
                    
            except Exception as e:
                # This catches the error if a month is missing (like August 2021 in the tutorial)
                print(f"Skipped {year}/{month:02d} - File likely does not exist.")
```
